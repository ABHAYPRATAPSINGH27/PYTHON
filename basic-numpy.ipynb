# Why numpy is used instead of lists 

NumPy is used instead of lists for several reasons:
1. **Performance**: NumPy arrays are more efficient for numerical operations compared to Python lists.
2. **Memory Usage**: NumPy arrays use less memory than Python lists.
3. **Functionality**: NumPy provides a wide range of mathematical functions that can be applied to arrays directly.
4. **Broadcasting**: NumPy supports broadcasting, which allows operations between arrays of different shapes.
5. **Integration**: NumPy integrates well with other scientific computing libraries in Python.
