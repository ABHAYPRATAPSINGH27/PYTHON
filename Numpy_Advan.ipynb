# Check it returns a Copy or View ?

```python
import numpy as np

n = np.arange(1, 14)
n[5] = 20
print(n)
id(n)
```

```python
import numpy as np
y = np.reshape(n, (1, 13)).base
print(y)
id(y)
```

```python
arr = np.array([1, 2, 3, 4, 5])
x=arr.view()
y = arr.copy()
arr[0] = 42

print(arr)
id(arr)
```

```python
print(x)
print(id(x))
print(x.base)
```

```python
print(y)
print(id(y))
print(y.base)
```

# Rotation of array

* Syntax -
  numpy.rot90(m,k=1,axes=(0,1))
  * m     Array of two or more dimensions
  * k     Number of times the array is rotated by 90 degrees
  * axes  The array is rotated in the plane defined by the axes 
  

```python
m = np.array([[1,2],[3,4],[5,6]])

```

```python
import numpy as np

m = np.array([[1, 2],[3, 4],[5, 6]])

result = np.rot90(m, k=4, axes=(0, 1))
print(result)
```

```python
m = np.rot90(m, k=4)
print(m)
```
