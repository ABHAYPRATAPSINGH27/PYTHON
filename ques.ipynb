```python
def linear_search(arr, target):
    for index, value in enumerate(arr):
        if value == target:
            return index
    return -1


def binary_search(arr, target):
    low, high = 0, len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return -1


numbers = [1, 3, 5, 7, 9, 11, 13]
target = 7

print("Linear search index:", linear_search(numbers, target))
print("Binary search index:", binary_search(numbers, target))
```

```python
# Calculate frequency of each character using a function in Python

def character_frequency(text):
    frequency = {}

    for char in text:
        frequency[char] = frequency.get(char, 0) + 1

    return frequency


character = "My name is Abhay"
frequency = character_frequency(character)

print(frequency)
```
