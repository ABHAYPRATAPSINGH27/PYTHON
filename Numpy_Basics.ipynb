# Choice() Method

-> Generate a random sample from a given 1-D array. 
* It is used to create a random sample from a given 1-D array.
* The random sample is generated based on the specified size and probability distribution.

```python
import numpy as np
rng= np.random.default_rng()
x=np.random.choice([1,2,3,4,5,6,7,8], size=10)
y=np.random.choice([1,2,3,4,5,6,7,8], size=(2,2))
z = rng.choice([1,2,3,4,5,6,7,8])
print(x)
print(y)
print(z)

```

```python
import numpy as np
a=np.array([1,2,3,4,5])
b = np.array([[6,7,8,9,10],[1,2,3,4,5]])
print(a[4])                                            # Accessing element at index 4 in array a
print(b[0,4])                                          # Accessing element at row 0, column 4 in array b
print(b[1][4])                                         # Accessing element at row 1, column 4 in array b
print(b[1:])                                           # Slicing rows from index 1 to the end
print(b[:,2])                                          # Slicing columns from index 2 to the end
print(b[0:2,1:4])                                      # Slicing rows from index 0 to 1 and columns from index 1 to 3
print(b[-1])                                           # Accessing the last row
print(b[-1:,-1:])                                      # Slicing the last row and last column
```
