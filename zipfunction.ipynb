```python
# Zip function in python is used to combine two or more iterables (like lists, tuples, etc.) into a single iterable. 
# It pairs elements from the input iterables based on their positions. 
# The resulting iterable contains tuples, where each tuple consists of elements from the corresponding positions of the input iterables.
# Example

names = ['Abhay', 'Aditya', 'Amit']
ages = [19,20,21]
zipped = list(zip(names, ages))
print(zipped)
```

```python
names = ['Abhay', 'Aditya', 'Amit']
scores = [90,85,95]

combined = zip(names, scores)

print(list(combined))
```

```python
a = [1, 2, 3]
b = ['x', 'y']
print(list(zip(a, b)))
```

```python
#unzipping (reverse operation of zip)
pairs = [('Abhay', 19), ('Aditya', 20), ('Amit', 21)]
names, ages = zip(*pairs)
print(names)
print(ages)
```

```python
from itertools import zip_longest

a = [1, 2, 3]
b = ['x', 'y']
print(list(zip_longest(a, b, fillvalue='-')))
```
